```python
import ast
```

```python
def convert(obj):
    lst=[]
    for i in ast.literal_eval(obj):
        lst.append(i['name'])
    return lst    
```

```python
movie['genres']=movie['genres'].apply(convert)
```

```python
movie['keywords']=movie['keywords'].apply(convert)
```

```python
movie.head(1)
```

```python
movie.isnull().sum()
```

```python
def convert2(obj):
    lst=[]
    counter=0
    for i in ast.literal_eval(obj):
        if counter!=3:
            lst.append(i['name'])
            counter+=1
        else: 
            break    
    return lst    
```

```python
movie['cast']=movie['cast'].apply(convert2)
```

```python
movie.head(1)
```

```python
def fetch_director(obj):
    lst=[]
    
    for i in ast.literal_eval(obj):
        if i['job']=='Director':
            lst.append(i['name'])
            break
                
    return lst   
```

```python
movie['crew']=movie['crew'].apply(fetch_director)
```

```python
movie.head(1)
```

```python
###Overview is string so we need to convert this into string
movie['overview']=movie['overview'].apply(lambda x:x.split())
```

```python
movie.head(1)
```

```python
#'Sam Worthington'
#'SamWorthington'
# we need to remove this space
movie['genres']=movie['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movie['cast']=movie['cast'].apply(lambda x:[i.replace(" ","") for i in x])
movie['crew']=movie['crew'].apply(lambda x:[i.replace(" ","") for i in x])
movie['keywords']=movie['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movie.head(1)
```

```python
movie['tag']=movie['overview']+movie['genres']+movie['keywords']+movie['cast']+movie['crew']
```

```python
movie.head(1)
```

```python
df=movie[['movie_id','title','tag']]
```

```python
df.head()
```

```python
# Converting to string from list
df['tag']=df['tag'].apply(lambda x:" ".join(x))
```

```python
df.head()
```

```python
df['tag']=df['tag'].apply(lambda x:x.lower())
```

```python
df.head()
```

```python
# now we use bag of word(bow) to convert this text to vector
# In this we will extract the text which are top 5000 in frequency 
# Then we make vector for each movie of 5000 dim
# so overall dim = (Size of dataset , 5000 )
# vector[i][j]=represt how much time that vector[i][j] come in that movie tag
# we will remove stopwords
```

```python
import nltk
from nltk.stem.porter import PorterStemmer
```

```python
ps=PorterStemmer()
```

```python
def stem(text):
    l=[]
    for i in text.split():
        l.append(ps.stem(i))
    return " ".join(l)    
```

```python
df['tag']=df['tag'].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=5000,stop_words='english')
```

```python
vector=cv.fit_transform(df['tag']).toarray()
```

```python
vector
```

```python
cv.get_feature_names_out()
# In this mant word like action actions etc are considered 2 but we need to consider them as 1 so we need to apply stemming
```

```python
# Now we need to make matrix of cosine simlarity b/w movie
from sklearn.metrics.pairwise import cosine_similarity
```

```python
cosine_similarity(vector).shape
```

```python
similarity=cosine_similarity(vector)
```

```python
similarity
```

```python
df.head()
```

```python
# Now we need to make function that will reccomend 5 movie
#In this funtion we go to the movie index then sort the array in descending order and return top 5 movie
def recommend(movie):
    movie_index=df[df['title']==movie].index[0]
    distances=similarity[movie_index]
    movie_list=sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]

    for i in movie_list:
        print(df.iloc[i[0]].title)


```

```python
recommend('Avatar')
```
